```python
#3. Temperature Converter
temp=eval(input("enter the temperature="))
unit=input("enter unit (C/F):")
if unit == "C":
  f=(temp * 9 / 5)+32
  print("Temperature in Fahrenheit:",f,"*F")
elif unit == "F":
  c=(temp - 32) * 5 / 9
  print("Temperature in Celsius:",c,"*C")
else:
 print("Invalid unit. Please enter C or F. in capital letter")
```

```python
#7. Dictionary – Student Marks Summary
student={}
for i in range(3):
  name=input("enter name=")
  marks=int(input("enter marks="))
  student[name]=marks
print(student)
total = sum(student.values())
print("total marks",total)

avg = total / len(student)
print("average marks",avg)

```

```python
#8. Dictionary – Word Frequency
sent=input("sentence=")
words=sent.split()
freq={}
for word in words:
  if word in freq:
    freq[word]+=1
  else:
    freq[word]=1
print("Word frequency:")
for word,count in freq.items():

    print(word, ":", count)
```

```python
#8. Functions – Check Prime
def checkprime(num):
  """check if the number is prime or not"""
  if num<=1:
    return False
  for i in range(2,num):
    if num%i==0:
      print("not prime")

  else:
      print(num,"is a prime number")
num=int(input("enter the number="))
checkprime(num)
```

```python
#9. Vowel Counter
def vowels(s):
  """count the number of vowels in a string"""
  count = 0
  for i in s:
        if i in "aeiouAEIOU":
            count = count + 1
  return count

s = input("Enter a string: ")
print("Number of vowels:", vowels(s))

```
